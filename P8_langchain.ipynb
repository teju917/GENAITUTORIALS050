```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
from langchain_huggingface import HuggingFacePipeline

model_name = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)

pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=300,
    do_sample=False,
    return_full_text=False
)

llm = HuggingFacePipeline(pipeline=pipe)

print("Legal AI model loaded!")
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate.from_template("""
You are a Legal Information Assistant.

Your responsibility is to answer ONLY questions related
to law and legal information.

Rules:
1. Answer only legal-related questions.
2. If the question is unrelated to law, respond:
   "Sorry, I can only answer legal-related questions."
3. Explain legal concepts in simple English.
4. Give relevant legal sections or acts only when you
   are confident they are correct.
5. Never invent laws or legal sections.
6. Mention the applicable country or jurisdiction.
7. Do not pretend to be a lawyer.
8. Explain that your response is general legal information,
   not professional legal advice.

User Question:
{question}

Answer:
""")

legal_agent = prompt | llm | StrOutputParser()


def ask_legal_agent(question):

    if not question.strip():
        return "Please enter a question."

    response = legal_agent.invoke({
        "question": question
    })

    return response


print("Legal AI Agent is ready!")
```

```python

import gradio as gr

def chatbot(question):
    return ask_legal_agent(question)

with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown("""
    # ⚖️ Legal AI Assistant

    Ask questions related to law and legal information.

    **Powered by LangChain and a local AI model**
    """)

    question = gr.Textbox(
        label="Enter your legal question",
        placeholder="Example: What are fundamental rights in India?",
        lines=3
    )

    button = gr.Button("Get Legal Information", variant="primary")

    answer = gr.Markdown()

    button.click(
        chatbot,
        inputs=question,
        outputs=answer
    )

demo.launch(share=True)
```
